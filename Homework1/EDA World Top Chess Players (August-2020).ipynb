```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_openml
from data_profiling import ProfileReport
```

```python
data = fetch_openml(
    name="World-Top-Chess-Players-(August-2020)",
    version=1,
    as_frame=True,
    read_csv_kwargs={"low_memory": False}
)

df = data.frame

df.shape
```

```python
# Построить EDA отчет и сохранить его в файл рядом с ноутбуком.

profile = ProfileReport(
    df,
    title="Chess Players EDA"
)

profile.to_file("Chess_EDA.html")
```
